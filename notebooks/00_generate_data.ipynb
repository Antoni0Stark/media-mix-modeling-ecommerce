```python
import pandas as pd
import numpy as np

np.random.seed(42)

weeks = pd.date_range(start="2022-01-01", peridos=104, freq="W")

data = pd.DataFrame({
    "date": weeks,
    "paid_search_spend": np.random.normal(50000, 8000, size=104),
    "paid_social_spend": np.random.normal(40000, 7000, size=104),
    "display_spend": np.random.normal(20000, 5000, size=104),
    "email_volume": np.random.normal(300000, 40000, size=104)
})

data.to_csv("../data/raw/marketing_data.csv", index=False)
data.head()
```
